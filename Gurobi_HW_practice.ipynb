```python
# Import gurobi package
from gurobipy import *
```

```python
# Name model
model=Model("Red_River")

# Define decision variables
xA={}
xA=model.addVar(vtype=GRB.CONTINUOUS, lb=2, ub=20)
xB={}
xB=model.addVar(vtype=GRB.CONTINUOUS, lb=3, ub=20)

# Define objective function
z=9*xA+13*xB
model.setObjective(z)
model.modelSense=GRB.MAXIMIZE
model.update()

# Add constraints
model.addConstr(2*xA+xB<=36)
model.addConstr(xA+3*xB<=42)
model.update()

# Solve model
model.optimize()

# Print outputs
if model.status==GRB.OPTIMAL:
  print("\n Optimal value:", model.objVal)
  print("Batch quantities")
  print("xA", xA.x)
  print("xB", xB.x)
```

```python
# Name model
model=Model("Investment_Portfolio")

# Define decision variables
xC={}
xC=model.addVar(vtype=GRB.CONTINUOUS, ub=GRB.INFINITY)
xD={}
xD=model.addVar(vtype=GRB.CONTINUOUS, lb=10, ub=GRB.INFINITY)

# Define objective function
z=xC+xD
model.setObjective(z)
model.modelSense=GRB.MINIMIZE
model.update()

# Add constraints
model.addConstr(0.4*xC+0.15*xD>=22)
model.addConstr(xC-1.8*xD<=0)
model.addConstr(xC+xD<=110)
model.update()

# Solve model
model.optimize()

# Print outputs
if model.status==GRB.OPTIMAL:
  print("\n Optimal value:", model.objVal)
  print("Investment Funds")
  print("xC", xC.x)
  print("xD", xD.x)
```

```python
# Homework 4 practice - dual problem
# Create new model
model = Model("Dual_Paint_Problem")
model.setParam('OutputFlag', 1) # Ensure output is shown

# Create variables
wM = model.addVar(vtype=GRB.CONTINUOUS, lb=0, name="wM")
wG = model.addVar(vtype=GRB.CONTINUOUS, lb=-GRB.INFINITY, ub=0, name="wG")

# Set objective function
model.setObjective(26 * wM + 8 * wG, GRB.MINIMIZE)

# Add constraints
model.addConstr(wM + wG >= 11, "Dual_C1")
model.addConstr(2 * wM + wG <= 13, "Dual_C2")

# Optimize model
model.optimize()

# Output results (only if an optimal solution is found)
if model.status == GRB.OPTIMAL:
    print("\n--- Optimal Solution for the Dual Problem ---")
    print(f"Optimal objective value (W_min): {model.ObjVal:.2f}")
    print("Optimal Dual Variables:")
    print(f"  wM: {wM.X:.2f}")
    print(f"  wG: {wG.X:.2f}")

    print("\n--- Primal Problem Insight (Duality Theorem) ---")
    # The optimal primal objective (Z_max) equals the optimal dual objective (W_min)
    print(f"Max Primal Profit (Z_max): {model.ObjVal:.2f}")
    print(f"Shadow Price for Primal Constraint 1: {wM.X:.2f}")
    print(f"Shadow Price for Primal Constraint 2: {wG.X:.2f}")

else:
    print(f"Optimization failed or status is non-optimal. Status code: {model.status}")
```
